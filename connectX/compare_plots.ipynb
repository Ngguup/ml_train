```python
import json
import matplotlib.pyplot as plt


labels = ['log2', 'log3']
logs_list = []

for label in labels:
    with open(f'logs/{label}.json', 'r') as f:
        logs_list.append(json.load(f))
```

```python
for key in ['loss', 'lr']:
    min_len = min(map(len, [logs[key] for logs in logs_list]))
    
    for logs in logs_list:
        logs[key] = logs[key][:min_len]
```

```python
fig, ax = plt.subplots()


for label, logs in zip(labels, logs_list):
    ax.plot(range(len(logs['loss'])), logs['loss'], label=label)

ax.legend()
ax.set_title('loss')
ax.set_xlabel('step')
ax.set_ylabel('loss')
plt.show()
```

```python
fig, ax = plt.subplots()


for label, logs in zip(labels, logs_list):
    ax.plot(range(len(logs['lr'])), logs['lr'], label=label)

ax.legend()
ax.set_title('lr')
ax.set_xlabel('step')
ax.set_ylabel('lr')
plt.show()
```

```python
fig, ax = plt.subplots()

for label, logs in zip(labels, logs_list):
    ax.plot(logs['val_step'], logs['val_reward'])

ax.set_title('validation reward')
ax.set_xlabel('step')
ax.set_ylabel('val_reward')
plt.show()
```
